```python
words = open("names.txt", "r").read().splitlines()
```

```python
words[:10]
train_split = words[:int(len(words)*0.8)]
test_split = words[int(len(words)*0.8):]
len(train_split)
len(test_split)
```

```python
vocab = ""

for word in words:
  vocab += word

vocab = sorted(list(set(vocab)))
vocab
```

```python
itos = {i+1:c for i, c in enumerate(vocab)}
itos[0] = "."
itos
```

```python
stoi = {c:i for i, c in itos.items()}
stoi
```

```python
import torch
```

```python
N_Train = torch.zeros((27,27,27))
N_Train

N_Test = torch.zeros((27,27,27))
N_Test
```

```python
for word in train_split:
  chs = ["."] + list(word) + ["."]
  for ch1, ch2, ch3 in zip(chs, chs[1:], chs[2:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    ix3 = stoi[ch3]
    N_Train[ix1][ix2][ix3] += 1

for word in test_split:
  chs = ["."] + list(word) + ["."]
  for ch1, ch2, ch3 in zip(chs, chs[1:], chs[2:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    ix3 = stoi[ch3]
    N_Test[ix1][ix2][ix3] += 1
```

```python
N_Train.shape
```

```python
P = (N_Train+2).float()
P /= P.sum(2, keepdims=True)
P.shape
```

```python
P.shape
```

```python
g = torch.Generator().manual_seed(2147483647)
for i in range(5):
  ix = 0
  jx = 0
  out = []
  while True:
    p = P[ix, jx]
    ix = jx
    jx = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[jx])
    if jx == 0:
      break
  name = "".join(out)
  print(name)
```

```python
log_likelihood = 0.0
n = 0

for word in test_split:
  chs = ["."] + list(word) + ["."]
  for ch1, ch2, ch3 in zip(chs, chs[1:], chs[2:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    ix3 = stoi[ch3]
    prob = P[ix1, ix2, ix3]
    logprob = torch.log(prob)
    log_likelihood += logprob
    n += 1

print(f"{log_likelihood=}")
print(f"{n=}")
nll = -log_likelihood
print(f"{nll=}")
print(f"{nll/n=}")
```
