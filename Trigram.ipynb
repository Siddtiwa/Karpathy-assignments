```python
import torch
N = torch.zeros((27,27,27), dtype=torch.int32)
N
```

```python
chars = sorted(list(set("".join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi["."] = 0
itos = {i:s for s,i in stoi.items()}
```

```python
stoi
```

```python
itos
```

```python
for word in words:
  chs = ["."] + list(word) + ["."]
  for ch1, ch2, ch3 in zip(chs, chs[1:], chs[2:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    ix3 = stoi[ch3]
    N[ix1, ix2, ix3] += 1
```

```python
N
```

```python
import matplotlib.pyplot as plt

%matplotlib inline

for k in range(27):

    plt.figure(figsize=(12, 12))

    plt.imshow(N[:, :, k], cmap='Blues')

    for i in range(27):
        for j in range(27):

            chstr = itos[i] + itos[j]

            plt.text(
                j, i,
                chstr,
                ha="center",
                va="bottom",
                color="gray"
            )

            plt.text(
                j, i,
                N[i, j, k].item(),
                ha="center",
                va="top",
                color="gray"
            )

    plt.title(f"Third character: '{itos[k]}'")
    plt.xlabel("Second character")
    plt.ylabel("First character")
    plt.axis('off')

    plt.show()
```

```python
P = (N+1).float()
P /= P.sum(2, keepdims=True)
```

```python
g = torch.Generator().manual_seed(2147483647)
ix = torch.multinomial(P[0,0], num_samples=1, replacement=True, generator=g).item()
itos[ix]
```

```python
for i in range(5):
  ix = 0
  jx = 0
  out = []
  while True:
    p = P[ix, jx]
    ix = jx
    jx = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[jx])
    if jx == 0:
      break
  name = "".join(out)
  print(name)
```

```python
log_likelihood = 0.0
n = 0

for word in words:
  chs = ["."] + list(word) + ["."]
  for ch1, ch2, ch3 in zip(chs, chs[1:], chs[2:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    ix3 = stoi[ch3]
    prob = P[ix1, ix2, ix3]
    logprob = torch.log(prob)
    log_likelihood += logprob
    n += 1

print(f"{log_likelihood=}")
print(f"{n=}")
nll = -log_likelihood
print(f"{nll=}")
print(f"{nll/n=}")
```
